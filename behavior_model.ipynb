```python
#importing necessary library
import numpy as np
import pandas as pd
```

```python
#importing data
data = pd.read_csv("/content/adjusted_simulated_patient_data_2000.csv")
data.head(20)
```

```python
#all feature
drop_col = ["Age","First Diagnosis Age","Flagged Behaviors"]
feature_all = data.drop(drop_col,axis=1)

#target column
label = data["Flagged Behaviors"]
```

```python
# function for splitting into train and test sets
def split_dataset(features,test_size):
  features_train, features_test, target_train, target_test \
  =train_test_split(features, label, test_size = test_size, random_state = 42, shuffle=False)
  return features_train, features_test, target_train, target_test
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
#neural network
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
y = to_categorical(label,num_classes=5)   #for multiclass classification

X_train, X_test, y_train, y_test = train_test_split(feature_all,y,test_size=0.3,random_state=42)

# Feature Scaling
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)

model = Sequential([
    Dense(64, activation='relu', input_dim=X_train.shape[1]),
    Dense(48, activation='relu'),
    Dense(24, activation='relu'),
    Dense(12, activation='relu'),
    Dense(5, activation='softmax')
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

model.fit(X_train, y_train, batch_size=32, epochs=50)

loss, accuracy = model.evaluate(X_test, y_test)
print(f"Test Accuracy: {accuracy * 100:.2f}%")
```

```python
#function for predicting the behavior kind
def predict_flagged_behavior(score):
  if score == 0:
    return "Bipolar Disorder"
  elif score ==1:
    return "ADHD"
  elif score==2:
    return "Depressive Order"
  elif score==3:
    return "Panic Disorder"
  elif score==4:
    return "Sensory Processing Disorder"
```
